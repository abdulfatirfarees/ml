### Interactive Calculator
Run the cell below to use the interactive calculator.

```python
import ipywidgets as widgets
from IPython.display import display

# Create widgets
num1 = widgets.FloatText(value=0.0, description='Number 1:')
num2 = widgets.FloatText(value=0.0, description='Number 2:')
operation = widgets.Dropdown(
    options=[('Add (+)', '+'), ('Subtract (-)', '-'), ('Multiply (*)', '*'), ('Divide (/)', '/')],
    value='+',
    description='Operation:',
)
button = widgets.Button(description='Calculate', button_style='success')
output = widgets.Output()

# Calculation logic
def calculate(b):
    with output:
        output.clear_output()
        n1 = num1.value
        n2 = num2.value
        op = operation.value

        if op == '+':
            result = n1 + n2
        elif op == '-':
            result = n1 - n2
        elif op == '*':
            result = n1 * n2
        elif op == '/':
            if n2 == 0:
                print("Error: Division by zero!")
                return
            result = n1 / n2

        print(f"Result: {n1} {op} {n2} = {result}")

button.on_click(calculate)

# Display GUI
display(num1, num2, operation, button, output)
```
